```python

import pandas as pd
import numpy as np
df = pd.DataFrame({
    'A': [10, 20, 30, 40, 50],
    'B': [11, 21, 31, 41, 51],
    'C': [12, 22, 32, 42, 52],
    'D': [13, 23, 33, 43, 53]
})
df.loc[0, 'A'] = np.nan
df.loc[1, 'B'] = np.nan
df.loc[2, 'C'] = np.nan
df.loc[3, 'D'] = np.nan
print("DataFrame:")
print(df)
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python

import pandas as pd
import numpy as np
df = pd.DataFrame({
    'A': [10, 20, 30, 40, 50],
    'B': [11, 21, 31, 41, 51],
    'C': [12, 22, 32, 42, 52],
    'D': [13, 23, 33, 43, 53]
})
df.loc[0, 'A'] = np.nan
df.loc[1, 'B'] = np.nan
df.loc[2, 'C'] = np.nan
df.loc[3, 'D'] = np.nan

print("Original DataFrame:")
print(df)
df_rows = df.dropna()
print("\nAfter dropna():")
print(df_rows)
print("Number of rows remaining:", len(df_rows))
df_columns = df.dropna(axis='columns')
print("\nAfter dropna(axis='columns'):")
print(df_columns)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Name': ['Harshith', np.nan, 'Rahul', 'Priya'],
    'Marks': [85, np.nan, np.nan, 90],
    'Age': [20, np.nan, 21, np.nan]
})
print("Original DataFrame:")
print(df)
df_cleaned = df.dropna(how='all')
print("\nDataFrame after dropna(how='all'):")
print(df_cleaned)
```

```python
import pandas as pd
data = {
    'department': ['CSE', 'CSE', 'AIML', 'AIML', 'ECE', 'ECE',
                   'CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'F', 'M', 'F', 'M', 'F', 'M'],
    'salary': [40000, 45000, 50000, 55000, 35000, 40000,
               42000, 52000, 38000, 43000, 57000, 36000]
}

df = pd.DataFrame(data)
print("Employee Data:")
print(df)
pivot = df.pivot_table(
    values='salary',
    index='gender',
    columns='department',
    aggfunc='mean'
)
print("\nAverage Salary by Gender and Department:")
print(pivot)
```
